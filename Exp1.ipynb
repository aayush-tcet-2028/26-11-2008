```python
# Theory

# Python libraries simplify the main steps of preparing and understanding data. NumPy stores
# numerical values in arrays and performs fast mathematical operations. Functions such as mean
# and max summarize the values in an array. Pandas represents tabular data using rows and named
# columns in a DataFrame.

# The head method displays a few records, while shape gives row and column counts. The info
# method shows column types and the number of non-missing values. The describe method summarizes
# numerical data using statistics such as mean and quartiles. Missing values should be checked
# because they can affect later analysis and training. Filtering selects records that satisfy a
# condition, such as marks above 75. Grouping combines records by a category and calculates a
# summary for each group.

# Matplotlib creates graphs that help explain the dataset visually. A bar chart compares marks
# across students, while a scatter plot compares hours and marks. These operations form part of
# exploratory data analysis before machine learning. The small student dataset is created inside
# the program, so no CSV file is required. A visible relationship in a graph does not by itself
# prove that one variable causes another.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Create a small student dataset
marks = np.array([65, 78, 90, 55, 82, 70, 95, 60])
df = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Hours': [2, 4, 6, 1, 5, 3, 7, 2],
    'Marks': marks,
    'Group': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B']
})
```

```python
# Explore the dataset using Pandas and NumPy
print("First five records:\n", df.head())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
print("Summary:\n", df.describe())
print("Missing values:\n", df.isnull().sum())
print("Average marks:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Students scoring above 75:\n", df[df['Marks'] > 75])
print("Group averages:\n", df.groupby('Group')['Marks'].mean())
```

```python
# Visualize marks and the relationship with study hours
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.bar(df['Student'], df['Marks'], color='skyblue')
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.subplot(1, 2, 2)
plt.scatter(df['Hours'], df['Marks'], color='green')
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.tight_layout()
plt.show()
```
